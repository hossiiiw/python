```python
import pandas as pd


df =pd.read_csv("./world_population.csv")
df.head(23)
df.tail(1) 
df.sample(10)
# df.info(verbose=False)
df.describe()
df.shape
df.columns
df.index
df["CCA3"].head(25).tail(10).sample(5).describe()

#subsetting
df[df.columns[0:3]]
df[df.columns[-3:]]
df[df.columns[-5:-2]]
df_cr = df[["Country" ,"Rank", "Density (per km²)"]]

#iloc
df.iloc[5:10]
# ======row === columns
df.iloc[5:10 , 0:2 ]
df.iloc[0:10 , -3:-1]
df.iloc[[2]]

#loc
df.loc[: , ["Country"]]
# df.iloc[0:2, 0:2]
df2=df.loc[(df["Rank"] > 150 ) & (df["Area (km²)"] > 5000)][["Country" , "Rank" , "Area (km²)"]]
df2.reset_index()


df[(df["Rank"] > 90) & (df["Rank"] < 150)][["Country"]].reset_index()
df[df["Country"].str.contains("A",case=False)][["Country"]].head(10)

#query
df.query('(Rank > 200)')[["Country"]]
num = 150
df.query('(Rank > @num) and (Rank < 200) ')[["Country"]].reset_index()
df["Rank"].mean()
df["Area (km²)"].mean()
df["Area (km²)"].min()
df["Rank"].min()
df["Area (km²)"].max()
df["Rank"].sum()
df["Area (km²)"].count()
df[["Rank" , "Area (km²)"]].sum()
df[["2015 Population","2020 Population" , "2022 Population"]].agg(["min" ,"max" , "mean" , "count" , "sum"])
df["Continent"].value_counts()
df["Continent"].unique()

#group by
df.groupby("Continent")[["Rank"]].agg(["min" , "max" , "mean" , "count"])
df.groupby("Country")[["Area (km²)" , "Rank"]].agg(["min" , "max" , "mean" , "sum" ]).head(16).reset_index()

df["new"] = df["Rank"]  / 2
df=df.assign(newAssign = df["Rank"] //2)
df[["Country","new" , "newAssign"]]

df.sort_values("Continent" ,ascending=False)[["Continent" ,"Country", "Rank" ,"new" , "newAssign"  ]].head().reset_index()
df.sort_values(["Continent","Rank"] , ascending=[False , True])[["Continent" ,"Country", "Rank" ,"new" , "newAssign"  ]].head()


#missing value

df[["Country" , "2015 Population","2020 Population" , "2022 Population"]].isna()
df[["Country" , "2015 Population","2020 Population" , "2022 Population"]].dropna()
df[["Country" , "2015 Population","2020 Population" , "2022 Population"]].fillna(df["2022 Population"].mean())

#merg
df_asia = df[df["Continent"] == "Asia"].copy()
df_africa = df[df["Continent"] == "Africa"].copy()

result = pd.concat([df_asia , df_africa] )
result[["Country","Rank"]].loc[(df["Rank"] >10) & (df["Continent"] == "Asia") ]


df_rate =pd.DataFrame( {
  "Continent" : ["Asia" , "Africa" , "Eroupe"],
  "Rate" :[100,20,30],
  "Rank" : [200 , 50 , 45]
})

# df_rate

df_final = pd.merge(df , df_rate , left_on=["Continent"] , right_on=["Continent"])
df_final.head()
 
df_final.to_csv("final_df.csv")




```
